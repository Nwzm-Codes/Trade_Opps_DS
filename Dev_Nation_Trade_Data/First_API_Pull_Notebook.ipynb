```python
#------------------------------------#
#------Create Network Component------#
#------------------------------------#

Nodes = []

#------CREATE NODES---
Countries_only = dfFinalData[['reporterCode', 'reporterDesc']]
#de-dupe countries
Countries_Dict = dict(zip(dfFinalData['reporterCode'], dfFinalData['reporterDesc']))

#Loop through the de-duped dict and create the labelled nodes dict 
for id, name in Countries_Dict.items():
    Nodes.append({
        "ID" : id,
        "Label": str(name)
    })



```

```python


# ---------CREATE Edges

Edges = []

#Pull out trade lane columns only
Trade_Routes = dfFinalData[['reporterCode', 'partnerCode','refPeriodId', 'cmdDesc', 'primaryValue','flowCode']]
Trade_Routes_df = pd.DataFrame(Trade_Routes)

# De-dupe trade lanes
Trade_Routes_df_dedupes = Trade_Routes_df.drop_duplicates()

for lanes in range(len(Trade_Routes_df_dedupes)):
    
        if Trade_Routes_df_dedupes.loc[lanes,'flowCode'] == 'X':
            Edges.append({
                'From_ID' : Trade_Routes_df_dedupes.loc[lanes,'reporterCode'],
                'To_ID'   : Trade_Routes_df_dedupes.loc[lanes,'partnerCode'],
                'Product_Type' : Trade_Routes_df_dedupes.loc[lanes,'cmdDesc'],
                'Value'  : Trade_Routes_df_dedupes.loc[lanes,'primaryValue']
            })
    
        else:
        
            continue
        
        
Edges       
```
